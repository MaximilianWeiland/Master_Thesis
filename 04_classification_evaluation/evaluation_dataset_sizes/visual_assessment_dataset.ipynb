```python
# libraries for data loading and plotting
import matplotlib.pyplot as plt
import json
```

```python
# load the data
with open("eval_results/results_non_augmented.json", "r") as f:
    results_non_augmented = json.load(f)

with open("eval_results/results_augmented.json", "r") as f:
    results_augmented = json.load(f)
```

```python
# plot the model performance for different data sizes
datasizes = [500, 1000, 1500, 2000, 2500, 3000, 3500, 4000, 4500, 5000]

# plot the trajectory
plt.figure(figsize=(12,8))
plt.plot(datasizes, results_non_augmented, marker='o', label="Non-Augmented Data")
plt.plot(datasizes, results_augmented, marker='o', label="Augmented Data")
plt.title("Cross-Validated Model Performance with Different Dataset Sizes")
plt.xlabel("Dataset Size")
plt.ylabel("Seqeval F1 Score")
plt.legend()
plt.grid(True)
plt.show()
```
